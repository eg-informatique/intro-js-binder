## Les Fonctions en JavaScript

Une fonction est un ensemble d'instructions JavaScript réutilisables. 

L'utilisation de fonctions favorise l'**encapsulation** et la **réutilisation** du code. 

Une fonction débute par une **signature** qui comporte les éléments suivants:
* le mot clef **function**
* le nom de la fonction suivant les conventions en vigueur (camelCase)
* les paramètres de la fonction entre parenthèses
* le corps de la fonction entre accolades `{}`
* une instruction **return** optionnelle pour retourner une valeur

Exemples de signature:

```javascript
// exemple 1: sans paramètres, sans valeur de retour
function hello() {
    console.log("Hello!");
}

// exemple 2: avec paramètres, sans valeur de retour
function displayMessage(message) {
    console.log(message);
}

// exemple 3: avec paramètres et valeur de retour
function add(a, b) {
    return a + b;
}

// exemple 4: fonction avec plusieurs paramètres
function calculateRectangleArea(width, height) {
    return width * height;
}
```

Il est à noter que les instructions qui constituent le corps d'une fonction doivent être placées entre accolades `{}` après la signature. 

Une fonction qui retourne une valeur se termine par une instruction **return** suivie de la valeur retournée. Les fonctions qui ne retournent pas explicitement de valeur retournent `undefined` par défaut.

JavaScript supporte également d'autres syntaxes pour déclarer des fonctions:

```javascript
// Fonction fléchée (arrow function) - syntaxe moderne
const multiply = (a, b) => {
    return a * b;
};

// Fonction fléchée avec retour implicite
const square = x => x * x;

// Fonction anonyme assignée à une variable
const divide = function(a, b) {
    return a / b;
};
```

JavaScript est un langage à typage dynamique, ce qui signifie que les types des variables sont déterminés au moment de l'exécution:

```javascript
// Les variables peuvent contenir différents types
let x = 1;          // number
x = "test";         // string
x = true;           // boolean
x = [1, 2, 3];      // array
x = {name: "John"}; // object
```
