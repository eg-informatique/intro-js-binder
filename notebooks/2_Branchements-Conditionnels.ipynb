# Bases du JavaScript - Branchements Conditionnels

Nous allons commencer par découvrir les [instructions de branchement conditionnel](https://developer.mozilla.org/fr/docs/Web/JavaScript/Guide/Control_flow_and_error_handling#les_instructions_conditionnelles).

Les branchements conditionnels sont des éléments essentiels d'un programme qui permettent d'orienter son exécution vers une certaine branche (i.e., série d'instructions) suivant les conditions qui sont rencontrées.

## Evaluation d'expressions booléenes

Pour déterminer dans quelle branche orienter l'exécution du programme, un branchement conditionnel commence par évaluer une expression booléenne qui est évaluée à `true` ou `false`. 

Déterminer à quelle valeur sont évaluées les expressions booléennes qui suivent, puis compléter les mots clef `null` par la valeur booléenne correspondante (i.e., `true` ou `false`). 

NB: La fonction `console.assert` permet de vérifier de manière automatisée qu'une assertion est vraie. Elle est utilisée pour implémenter des tests automatisés en JavaScript pour vérifier par exemple que dans un contexte spécifique, une fonction retourne bien les valeurs attendues. 

Dans le cas où une assertion est vérifiée comme étant vraie, la fonction `console.assert` n'affiche rien. Dans le cas contraire, elle affiche un message d'erreur dans la console.

```typescript
// true
console.assert(true === null);

// false
console.assert(false === null);

// 2 > 3
console.assert((2 > 3) === null);

// 6 <= 7
console.assert((6 <= 7) === null);

// 3 == 3
console.assert((3 === 3) === null);

// 4 != 4
console.assert((4 !== 4) === null);

// true || false
console.assert((true || false) === null);

// true && false
console.assert((true && false) === null);

// 2 > 3 || 4 < 5
console.assert((2 > 3 || 4 < 5) === null);

// 2 > 3 && 4 < 5
console.assert((2 > 3 && 4 < 5) === null);
```
