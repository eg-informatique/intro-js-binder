### Qu'est qu'une variable?

En informatique, une variable est un symbole qui associe un nom appelé **identifiant** à une valeur. Les variables peuvent généralement changer de valeur au cours du temps, et possèdent un type (e.g., une chaîne de caractères (`string`), un nombre entier (`number`), ou un nombre à virgule (`number`)).

**Déclarer** une variable consiste à lui donner un nom et à indiquer son type. **Instancier** une variable consiste à lui donner une valeur.

Pour afficher la valeur d'une variable, on peut utiliser la commande "console.log". Par exemple, `console.log(x)` affiche la valeur de la variable `x`. Il est aussi possible d'afficher un message plus élaboré qui intègre la valeur de la variable `x` en utilisant l'une des commandes suivantes: 
* `console.log("La valeur de la variable est:" + x)`
* `console.log("La valeur de la variable est:", x)`
* `` console.log(`La valeur de la variable est: ${x}`)``

On souhaite afficher ceci:

```
La variable 'a' vaut: 8
La variable 'b' vaut: 3.14
La variable 'c' vaut: Bonjour
La variable 'd' vaut: true
```

### Variables de type tableau (`Array`)

Un tableau est une variable pouvant contenir un ensemble de valeurs stockées dans un ordre précis et qui peuvent comporter des doublons, comme par exemple un ensemble de chaînes de caractères (`string`). 

En JavaScript, il est conseillé de n'ajouter que des valeurs d'un même type à un tableau. 

La propriété `length` permet de déterminer la longuer d'un tableau (i.e., le nombre de ses éléments). Par exemple, `a.length` nous donne la longueur du tableau `a`.

La méthode `push` permet d'ajouter un élément à la fin d'un tableau. Par exemple, `a.push(3)` ajoute l'élément 3 au tableau `a`.

La méthode `pop` permet d'enlever le dernier élément du tableau. Par exemple, `a.pop()` enlève le dernier élément du tableau `a`.

La méthode `splice` permet d'enlever un élément à un indice spécifique. Par exemple, `a.splice(3, 1)` enlève le 4ème élément de `a`.

### Variables de type ensemble (`Set`)

Un ensemble est une collection de valeurs stockées sans ordre précis et qui ne peuvent comporter de doublons.
