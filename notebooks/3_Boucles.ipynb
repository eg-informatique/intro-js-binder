# Bases du JavaScript - Boucles

Une boucle est un élément essentiel en programmation qui permet de répéter un ensemble d'instructions tant qu'une certaine condition est vérifiée. Chaque passage dans la boucle est appelée une **itération**.

## Boucle `while`

Une boucle `while` a la structure suivante:

```javascript
while (condition) {
    instruction1;
    instruction2;
    ...
}
```

Avec ce type de boucle, il faut veiller à ne pas entrer dans une boucle infinie. Un exemple simple est la boucle `while` suivante qui affiche le message 'Bonjour' à l'infini.

```javascript
while (true) {
    console.log("Bonjour");
}
```

Une boucle `while` utilise souvent un **incrément de boucle** qu'il convient d'incrémenter ou de décrémenter à chaque itération afin de ne pas risquer d'entrer dans une boucle infinie.

L'exemple suivant utilise un incrément de boucle `i` et affiche 3 fois le message "Bonjour":

```javascript
let i = 2;

while (i >= 0) {
    console.log("Bonjour");
    i = i - 1;
}
```

## Boucle `for` (éléments)

Une boucle `for` peut être utilisée pour parcourir **directement** les éléments d'un tableau ou d'une chaîne de caractères.

Le code suivant affiche les éléments du tableau `monTableau` un par un:

```javascript
for (let element of monTableau) {
    console.log(element);
}
```

## Boucle `for` (indices)

Une boucle `for` peut être utilisée pour parcourir une **séquence** d'entiers servant à indexer par exemple les éléments d'un tableau.

Une boucle `for` classique en JavaScript utilise la structure suivante pour créer une séquence d'entiers:

* `for (let i = 0; i < 10; i++)` crée une séquence d'entiers de 0 à 9
* `for (let i = 1; i < 10; i++)` crée une séquence d'entiers de 1 à 9
* `for (let i = 1; i < 10; i += 2)` crée une séquence d'entiers de 1 à 9 par pas de 2 (i.e., 1,3,5,7,9)

Le code suivant affiche les entiers 1, 3, 5, 7 et 9:

```javascript
for (let i = 1; i < 10; i += 2) {
    console.log(i);
}
```

NB: Contrairement aux boucles `for` précédentes qui accèdent directement aux éléments d'un tableau, la boucle ci-dessus utilise un **incrément de boucle** qui parcourt une séquence d'entiers. Il est commun d'appeler cet incrément de boucle `i` car il est souvent utilisé comme un compteur ou pour indexer les éléments d'un tableau.
